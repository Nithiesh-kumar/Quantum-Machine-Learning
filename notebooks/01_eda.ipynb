# 01. Exploratory Data Analysis (EDA) - Patient Health Risk Prediction
**Project:** Quantum Machine Learning System for Patient Health Risk Prediction  
**Dataset:** UCI Heart Disease (Cleveland Clinic)  
**Target:** `num` (Binarized: 0 = Low Risk / Absence, 1 = High Risk / Presence)  

> **Medical Disclaimer**: This notebook is part of an educational prototype. Not for clinical diagnosis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.data.loader import load_dataset

df = load_dataset('heart')
print('Dataset Shape:', df.shape)
df.head()
```

## 1. Data Types & Missing Value Audit

```python
audit = pd.DataFrame({
    'Dtype': df.dtypes,
    'Missing': df.isnull().sum(),
    'Missing (%)': (df.isnull().sum() / len(df) * 100).round(2)
})
audit
```

## 2. Target Class Balance

```python
print(df['num'].value_counts())
sns.countplot(data=df, x='num', palette=['#2b5c8f', '#d95f02'])
plt.title('Heart Disease Risk (0: Low, 1: High)')
plt.show()
```

## 3. Clinical Feature Correlations

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='vlag', center=0)
plt.title('Correlation Matrix')
plt.show()
```

## 4. Key Clinical Insights
- **Balance**: 54.1% low risk vs 45.9% high risk provides a robust baseline.
- **Physiological Drivers**: Exercise induced angina (`exang`), maximum heart rate (`thalach`), and ST depression (`oldpeak`) are the strongest clinical predictors.
- **Quantum Readiness**: Dimension reduction (PCA / SelectKBest to $\le 8$ qubits) with angle scaling $[0, \pi]$ preserves key variance while respecting qubit constraints.
